# Estructura base de un sistema difuso

```text
                                      Base de reglas
                                            |
                                            V
Entradas -> Modulo de entradas -> Motor de inferencia -> Modulo de salidas
```

Para tener un sistema difuso vamos a tomar como referencia un sistema de control para un calentador de agua. Con el sistema difuso como linea de retroalimentación.

## Variables linguisticas

Cuadrupla que tiene $VL = \{n, U_n, T_n, S_n\}$ con:

- $n$: Nombre de la variable
- $U_n$: Universo al cual pertenece la variable (Rango de uso en este caso de la temperatura; 20 a 100)
- $T_n$: Terminos linguisticos en los que se maneja la variable  (Bajo, Medio, Alto)
- $S_n$: Significador de la variable (Conjuntos difusos que permiten interpretar las proposiciones)

## Base de reglas

**R1:** Si la temperatura es baja, entonces el gas se pone en alto. `t es B -> g es A`

**R2:** Si la temperatura es media, entonces el gas se pone en medio. `t es M -> g es M`

**R3:** Si la temperatura es alta, entonces el gas se pone en bajo. `t es A -> g es B`
